```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt 
```

```python
df_1_2 = pd.read_csv('consistency_experiment_a_1_decomp_lvl_2.csv')
df_1_3 = pd.read_csv('consistency_experiment_a_1_decomp_lvl_3.csv')
df_1_6 = pd.read_csv('consistency_experiment_a_1_decomp_lvl_6.csv')
df_5_2 = pd.read_csv('consistency_experiment_a_5_decomp_lvl_2.csv')
df_5_3 = pd.read_csv('consistency_experiment_a_5_decomp_lvl_3.csv')
df_5_6 = pd.read_csv('consistency_experiment_a_5_decomp_lvl_6.csv')
df_25_2 = pd.read_csv('consistency_experiment_a_25_decomp_lvl_2.csv')
df_25_3 = pd.read_csv('consistency_experiment_a_25_decomp_lvl_3.csv')
df_25_6 = pd.read_csv('consistency_experiment_a_25_decomp_lvl_6.csv')
```

```python
#MSE
a_curr = 5
decomp_lvl_curr = 2
fig_5_2 = plt.figure()
graph_5_2 = sns.boxplot(data = df_5_2, x = 'N', y = 'MSE', hue = 'Method')
plt.title('MSE: a = ' +str(a_curr)+ ', Number of Clusters = ' + str(decomp_lvl_curr))
plt.ylim(0, 1)
fig_5_2.savefig('MSE_a_'+str(a_curr)+'_decomp_lvl_'+str(decomp_lvl_curr)+'_resize.png',bbox_inches = 'tight')
```
